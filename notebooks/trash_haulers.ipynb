```python
hauler = pd.read_csv('../data/trash_hauler_report.csv')
```

```python
hauler.rename(columns={'Request ': 'Complaint'}, inplace=True)
```

```python
hauler = hauler.dropna(subset=['Complaint', 'Description'])
```

dropna function removes null/missing values.
subset=['Incident Address'] tells Pandas to check only the Incident Address column.




```python
hauler['Complaint'] = hauler['Complaint'].astype(str)
```

```python
hauler['Description'] = hauler['Description'].astype(str)
```

```python
missed_pickups = hauler.loc[
    hauler['Complaint'].str.contains('miss', case=False, na=False) |
    hauler['Description'].str.contains('miss', case=False, na=False)
]
```

The str.contains function checks to see if 'miss' exists anywhere within Complaint and Description. The false parameter makes capitalization case not important or null.

```python
missed_pickups = missed_pickups.dropna(subset=['Trash Hauler'])
```

```python
missed_pickups
```

```python
trash_hauler = missed_pickups.loc[
    missed_pickups['Trash Hauler'].str.contains("RED RIVER", case=False, na=False)
]
```

```python
trash_hauler = trash_hauler.dropna(subset=['Incident Address'])
```

```python
address_counts = trash_hauler['Incident Address'].value_counts()
address_counts
```

```python
frequent_addresses = address_counts[address_counts >= 2]
frequent_addresses
```

```python
.index brings back only unique addresses that meet the condition of appearing at least twice. 
address_counts pulls the actual addresses, not the count after filtering.
```

```python
trash_hauler = trash_hauler[trash_hauler['Incident Address'].isin(frequent_addresses)]
```

Filters trash_hauler DF to only include the rows where incident address is one of the frequent addresses (those that appeared 2 or more times).
